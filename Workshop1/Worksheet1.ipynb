```python
#3
A = np.array([[3, 4], [7, 8]])
B = np.array([[5, 3], [2, 1]])

AB = np.dot(A,B)

left = AB.T

aFlip = A.T
bFlip = B.T

right  = np.dot(bFlip,aFlip)
print(left)

print(right)
```
